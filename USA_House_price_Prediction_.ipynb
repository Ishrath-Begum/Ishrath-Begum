**Model Building and Evaluation**

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X=df.drop(columns=['Price','Address'])
y=df['Price']
```

**Observations**

* Features (X):

All columns except Price (target variable) and Address (non-numeric, irrelevant for prediction) are included.
This ensures the model uses only meaningful numeric predictors: income, house age, rooms, bedrooms, and population.
* Target (y):

The Price column is set as the dependent variable.

This is the value the regression model will learn to predict based on the selected features.

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

**Obseravtions**

* Dataset split into 70% training and 30% testing.

* Training set used to fit the model; testing set used to check performance on unseen data.

* Random state = 42 ensures reproducibility.

* Conclusion: This step prevents overfitting and allows unbiased evaluation of the regression model.

```python
X_train.shape
```

```python
y_train.shape
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_train_pred=model.predict(X_train)
y_test_pred=model.predict(X_test)
```

**Model Prediction**

* Model predicted prices for both training and testing sets.

* Training predictions check how well the model learned.

* Testing predictions show how well it generalizes to unseen data.

* Conclusion: Predictions are ready for evaluation using metrics (MAE, MSE, R²).

```python
y_test_pred
```

```python
plt.scatter(y_train,y_train_pred,color='blue',label='Training data')
plt.scatter(y_test,y_test_pred,color='red',label='Testing data')
min_value= min(y_test.min(),y_train.min())
max_value=max(y_test.max(),y_train.max())
plt.plot([min_value,max_value],[min_value,max_value],color='black',linestyle='--',label='perfect prediction')
plt.xlabel('Actual House Price')
plt.ylabel('Predicted House Price')
plt.title('Actual vs. Predicted House Prices')
plt.legend()
plt.show()
```

**Observations**

* Training (blue) and testing (red) points cluster closely around the perfect prediction line.

* Shows the model learned patterns well and generalizes effectively.

* Minimal deviation indicates high accuracy and low error.

Conclusion: The regression model performs reliably on both training and unseen test data.

```python
mae = mean_absolute_error(y_test, y_test_pred)
mse = mean_squared_error(y_test, y_test_pred)
r2 = r2_score(y_test, y_test_pred)
```

```python
print("MAE:", mae)
print("MSE:", mse)
print("R² Score:", r2)
```

**Model Evaluation Metrics**
* Mean Absolute Error (MAE): 81,135  
→ On average, predictions are off by about ₹81K.

* Mean Squared Error (MSE): 10,068,422,551  
→ Larger errors are penalized more heavily, but still within acceptable range.

* R² Score: 0.91  
→ The model explains 91% of the variance in house prices, showing strong predictive power.

Conclusion
The regression model performs very well — low error values and a high R² score confirm that it generalizes effectively to unseen data.

**Final Conclusion**

*  The House Price Prediction model using Linear Regression achieved strong results, with an MAE of ~81K, MSE of ~1.0×10¹⁰, and an R² score of 0.91. These metrics confirm that the model explains over 90% of the variance in house prices and makes predictions with relatively low error.

* The scatter plot of actual vs. predicted values shows that both training and testing data points align closely with the perfect prediction line, proving the model generalizes well to unseen data.
